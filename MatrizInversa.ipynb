```python
import numpy as np
#Se pide al usuario el número de columnas y número de filas que tendrá la matriz
m = int(input('Dame el número de columnas m para la matriz: '))
```

```python
#Se pide al usuario las entradas de la matriz
matriz = np.zeros((m,m))
for i in range(m):
  for j in range(m):
    matriz[i][j] = float(input('Elemento (%d,%d): ' % (i, j)))

print( "su primer matriz es:")
print(matriz)
```

```python
#Funcion auxiliar para imprimir la matriz aumentada [A | I] con division y sin puntos sueltos
def imprimir_aumentada(M, n):
    for fila in M:
        #Formateamos la parte izquierda (Matriz A) y la parte derecha (Identidad/Inversa)
        izquierda = " ".join([f"{val:8.4f}" for val in fila[:n]])
        derecha = " ".join([f"{val:8.4f}" for val in fila[n:]])
        print(f"[ {izquierda}  |  {derecha} ]")

#Funcion auxiliar para imprimir una matriz normal sin puntos sueltos
def imprimir_matriz(M):
    for fila in M:
        elementos = " ".join([f"{val:8.4f}" for val in fila])
        print(f"[ {elementos} ]")

#Calculamos la inversa de la matriz paso a paso mediante el metodo de Gauss-Jordan
def calcular_inversa(A):
    n = A.shape[0]
    I = np.eye(n) #Creamos la matriz identidad de tamano n x n

    #Unimos la matriz original y la identidad para formar la matriz aumentada [A | I]
    aumentada = np.hstack((A.astype(float).copy(), I))
    aumentada[np.isclose(aumentada, 0.0)] = 0.0
    paso = 1

    print("\nMatriz aumentada inicial [A | I]:")
    imprimir_aumentada(aumentada, n)

    for i in range(n):
        #Validamos si el pivote es cero para intercambiar filas
        if np.isclose(aumentada[i, i], 0.0):
            cambio = False
            for k in range(i + 1, n):
                if not np.isclose(aumentada[k, i], 0.0):
                    aumentada[[i, k]] = aumentada[[k, i]]
                    aumentada[np.isclose(aumentada, 0.0)] = 0.0
                    cambio = True
                    print(f"\nPaso {paso}: Intercambio de Fila {i} <-> Fila {k}")
                    imprimir_aumentada(aumentada, n)
                    paso += 1
                    break

            #Si no se pudo encontrar un pivote distinto de cero, la matriz no tiene inversa
            if not cambio:
                print(f"\nPaso {paso}: El pivote en la posicion ({i},{i}) es cero y no hay filas debajo para intercambiar.")
                print("La matriz NO tiene inversa porque su determinante es 0 (es una matriz singular).")
                return None

        #Dividimos la fila entre el pivote para convertirlo en 1
        pivote = aumentada[i, i]
        if not np.isclose(pivote, 1.0):
            aumentada[i] = aumentada[i] / pivote
            aumentada[np.isclose(aumentada, 0.0)] = 0.0 #Evitamos el -0.0000 convirtiendolo en 0.0 positivo
            print(f"\nPaso {paso}: Fila {i} = Fila {i} / ({pivote:.4f})")
            imprimir_aumentada(aumentada, n)
            paso += 1

        #Hacemos ceros en toda la columna i (arriba y abajo del pivote)
        for j in range(n):
            if i != j and not np.isclose(aumentada[j, i], 0.0):
                factor = aumentada[j, i]
                aumentada[j] = aumentada[j] - (factor * aumentada[i])
                aumentada[np.isclose(aumentada, 0.0)] = 0.0 #Evitamos el -0.0000 convirtiendolo en 0.0 positivo
                print(f"\nPaso {paso}: Fila {j} = Fila {j} - ({factor:.4f}) * Fila {i}")
                imprimir_aumentada(aumentada, n)
                paso += 1

    #Extraemos el lado derecho de la matriz aumentada que contiene la inversa
    inversa = aumentada[:, n:]
    return inversa # Devuelve la matriz inversa calculada

#Ejecutamos la funcion con la matriz ingresada
print("Método de Gauss-Jordan para Matriz Inversa")
matriz_inversa = calcular_inversa(matriz)

if matriz_inversa is not None:
    print("\nLa matriz inversa es:")
    imprimir_matriz(matriz_inversa)
```
