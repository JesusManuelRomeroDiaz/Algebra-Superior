```python
import numpy as np
#Se pide al usuario el número de columnas y número de filas que tendrá la matriz
m = int(input('Dame el número de columnas m para la matriz: '))
```

```python
#Se pide al usuario las entradas de la matriz
matriz = np.zeros((m,m))
for i in range(m):
		for j in range(m):
			matriz[i][j] = float(input('Elemento (%d,%d): ' % (i, j)))

print( "su primer matriz es:")
print(matriz)
```

```python
# Calculamos el determinante paso a paso mediante una funcion recursiva de expansion por cofactores
def determinante_cofactores(A):
    n = A.shape[0]

    #Caso base 1x1
    if n == 1:
        return 0.0  if np.isclose(A[0, 0], 0.0) else A[0, 0]
                    #Este if nos ayuda a validar si el resultado es prácticamente cero

    #Caso base 2x2
    if n == 2:
        det_2x2 = (A[0, 0] * A[1, 1]) - (A[0, 1] * A[1, 0])
        if np.isclose(det_2x2, 0.0): #Validamos si el resultado es prácticamente cero
            det_2x2 = 0.0
        print(f"\nDet 2x2: ({A[0,0]:.2f} * {A[1,1]:.2f}) - ({A[0,1]:.2f} * {A[1,0]:.2f}) = {det_2x2:.4f}")
        return det_2x2

    det_total = 0.0
    terminos_str = [] #Esto deja vacio el arreglo para guardar los resultados en formato de texto

    print(f"Expandiendo matriz de {n}x{n} por la primera fila:")
    for j in range(n):
        signo = (-1) ** j #Eso es para que los signos se alternen según la posición es decir (-1)^(0+j)
        simbolo_signo = "+" if signo > 0 else "-"
        elemento = 0.0 if np.isclose(A[0, j], 0.0) else A[0, j]

        #Creamos la submatriz y eliminamos fila 0 y columna j
        submatriz = np.delete(np.delete(A, 0, axis=0), j, axis=1)

        print(f"\nElemento (0,{j}) = {elemento:.2f} | Signo de posicion: ({simbolo_signo})")
        print(f"Submatriz menor M_0{j}:\n{submatriz}")

        #Llamada recursiva para el cálculo de cofactor
        det_sub = determinante_cofactores(submatriz)
        cofactor = signo * elemento * det_sub

        #Evitamos el -0.0000 convirtiendolo en 0.0 positivo
        if np.isclose(cofactor, 0.0):
            cofactor = 0.0

        det_total += cofactor

        #Imprimimos la operacion y el resultado del cofactor actual
        print(f"Termino {j+1}: ({simbolo_signo}1) * ({elemento:.2f}) * ({det_sub:.4f}) = {cofactor:.4f}")
        terminos_str.append(f"({cofactor:.4f})") #append guarda cada cofactor en el arreglo

    if np.isclose(det_total, 0.0):
        det_total = 0.0

    #Tomamos los textos guardados en el arreglo del inicio y los une con signo para mostrar la suma final
    print(f"\nSuma de cofactores ({n}x{n}): " + " + ".join(terminos_str) + f" = {det_total:.4f}")
    return det_total #Devuelve el determinante calculado

#Ejecutamos la funcion con la matriz ingresada
print("Método de Cofactores")
determinante = determinante_cofactores(matriz)
print(f"\nEl determinante de la matriz es: {determinante:.4f}")
```
