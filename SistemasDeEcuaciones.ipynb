```python
import numpy as np
#Se pide al usuario el número de incógnitas y ecuaciones que tendrá el sistema
m = int(input('Dame el número de incógnitas m para el sistema: '))
```

```python
#Se pide al usuario las entradas de la matriz de coeficientes y el vector de resultados
matriz = np.zeros((m,m))
b = np.zeros((m,1))

print("Ingresa los coeficientes de la matriz A:")
for i in range(m):
  for j in range(m):
    matriz[i][j] = float(input('Elemento (%d,%d): ' % (i, j)))

print("\nIngresa los términos independientes del vector b:")
for i in range(m):
  b[i][0] = float(input('Resultado de la ecuación %d: ' % (i)))

print("\nsu matriz de coeficientes es:")
print(matriz)
print("\nsu vector de resultados es:")
print(b)
```

```python
#Configuramos NumPy para mostrar siempre 4 decimales fijos
np.set_printoptions(precision=4, floatmode='fixed', suppress=True)

#Funcion auxiliar para imprimir la matriz aumentada [A | b] con division y ordenada
def imprimir_sistema(M, n):
    for fila in M:
        #Formateamos las primeras n columnas (Matriz A) y la ultima columna (Vector b)
        coeficientes = " ".join([f"{val:8.4f}" for val in fila[:n]])
        resultado = f"{fila[n]:8.4f}"
        print(f"[ {coeficientes}  |  {resultado} ]")

#Resolvemos el sistema de ecuaciones paso a paso mediante Eliminacion Gaussiana
def eliminacion_gaussiana(A, vector_b):
    n = A.shape[0]
    #Unimos la matriz A y el vector b para formar la matriz aumentada [A | b]
    aumentada = np.hstack((A.astype(float).copy(), vector_b.astype(float).copy()))
    aumentada[np.isclose(aumentada, 0.0)] = 0.0
    paso = 1

    print("\nMatriz aumentada inicial [A | b]:")
    imprimir_sistema(aumentada, n)

    #Eliminacion hacia adelante para formar una matriz triangular superior
    for i in range(n):
        #Validamos si el pivote es cero para intercambiar filas
        if np.isclose(aumentada[i, i], 0.0):
            cambio = False
            for k in range(i + 1, n):
                if not np.isclose(aumentada[k, i], 0.0):
                    aumentada[[i, k]] = aumentada[[k, i]]
                    aumentada[np.isclose(aumentada, 0.0)] = 0.0
                    cambio = True
                    print(f"\nPaso {paso}: Intercambio de Fila {i+1} <-> Fila {k+1}")
                    imprimir_sistema(aumentada, n)
                    paso += 1
                    break

            if not cambio:
                print(f"\nEl pivote en ({i+1},{i+1}) es cero. El sistema no tiene solucion unica.")
                return None

        #Hacemos ceros debajo del pivote actual
        for j in range(i + 1, n):
            if not np.isclose(aumentada[j, i], 0.0):
                factor = aumentada[j, i] / aumentada[i, i]
                aumentada[j] = aumentada[j] - (factor * aumentada[i])
                aumentada[np.isclose(aumentada, 0.0)] = 0.0 # Evitamos el -0.0000 convirtiendolo en 0.0 positivo
                print(f"\nPaso {paso}: Fila {j+1} = Fila {j+1} - ({factor:.4f}) * Fila {i+1}")
                imprimir_sistema(aumentada, n)
                paso += 1

    #Sustitucion hacia atras para despejar cada incognita
    print("\nSustitucion hacia atras (despeje de incognitas):")
    x = np.zeros(n)

    #Recorremos desde la ultima fila (n-1) hasta la primera (0)
    for i in range(n - 1, -1, -1):
        suma_conocidos = np.dot(aumentada[i, i+1:n], x[i+1:n])
        x[i] = (aumentada[i, n] - suma_conocidos) / aumentada[i, i]

        #Evitamos el -0.0000 convirtiendolo en 0.0 positivo
        if np.isclose(x[i], 0.0):
            x[i] = 0.0

        print(f"x_{i+1} = ({aumentada[i, n]:.4f} - ({suma_conocidos:.4f})) / ({aumentada[i, i]:.4f}) = {x[i]:.4f}")

    return x #Devuelve el arreglo con las soluciones del sistema

#Ejecutamos la funcion con la matriz y el vector ingresados
print("Método de Eliminación Gaussiana")
solucion = eliminacion_gaussiana(matriz, b)

if solucion is not None:
    print("\nLa solución del sistema de ecuaciones es:")
    for i in range(len(solucion)):
        print(f"x_{i+1} = {solucion[i]:.4f}")
```
