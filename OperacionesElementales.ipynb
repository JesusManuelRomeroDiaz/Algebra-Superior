```python
import numpy as np
#Se pide al usuario el número de columnas y número de filas que tendrá la matriz
m = int(input('Dame el número de columnas m para la matriz: '))
```

```python
#Se pide al usuario las entradas de la matriz
matriz = np.zeros((m,m))
for i in range(m):
  for j in range(m):
    matriz[i][j] = float(input('Elemento (%d,%d): ' % (i, j)))

print( "su primer matriz es:")
print(matriz)
```

```python
#Configuramos NumPy para mostrar siempre 4 decimales fijos
np.set_printoptions(precision=4, floatmode='fixed', suppress=True)

#Aplicamos las 3 operaciones elementales paso a paso para escalonar la matriz
def operaciones_elementales(A):
    n = A.shape[0]
    M = A.astype(float).copy() #Creamos una copia para trabajar sobre ella
    paso = 1

    for i in range(n):
        #Intercambio de filas (Fi <-> Fj) si el pivote es cero
        if np.isclose(M[i, i], 0.0):
            cambio = False
            for k in range(i + 1, n):
                if not np.isclose(M[k, i], 0.0):
                    M[[i, k]] = M[[k, i]]
                    M[np.isclose(M, 0.0)] = 0.0 #Evitamos el -0.0000 convirtiendolo en 0.0 positivo
                    cambio = True
                    print(f"\nPaso {paso} [Intercambio]: Fila {i+1} <-> Fila {k+1}")
                    print(np.round(M, 4))
                    paso += 1
                    break
            if not cambio:
                continue #Si toda la columna es cero, pasamos a la siguiente columna

        #Multiplicar una fila por un escalar no nulo (hacemos 1 el pivote)
        pivote = M[i, i]
        if not np.isclose(pivote, 1.0):
            escalar = 1.0 / pivote
            M[i] = escalar * M[i]
            M[np.isclose(M, 0.0)] = 0.0 #Evitamos el -0.0000 convirtiendolo en 0.0 positivo
            print(f"\nPaso {paso} [Multiplicacion por escalar]: Fila {i+1} = ({escalar:.4f}) * Fila {i+1}")
            print(np.round(M, 4))
            paso += 1

        #Sumar a una fila el multiplo de otra (hacemos ceros debajo del pivote)
        for j in range(i + 1, n):
            if not np.isclose(M[j, i], 0.0):
                factor = M[j, i]
                M[j] = M[j] - (factor * M[i])
                M[np.isclose(M, 0.0)] = 0.0 #Evitamos el -0.0000 convirtiendolo en 0.0 positivo
                print(f"\nPaso {paso} [Suma de multiplo de fila]: Fila {j+1} = Fila {j+1} - ({factor:.4f}) * Fila {i+1}")
                print(np.round(M, 4))
                paso += 1

    return M #Devuelve la matriz escalonada

#Ejecutamos la funcion con la matriz ingresada
print("Operaciones Elementales en una Matriz")
matriz_escalonada = operaciones_elementales(matriz)
print("\nLa matriz resultante después de las operaciones elementales es:")
print(np.round(matriz_escalonada, 4))
```
